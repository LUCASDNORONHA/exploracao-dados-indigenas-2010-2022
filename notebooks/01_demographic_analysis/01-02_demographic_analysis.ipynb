## Introdução

O Estudo 1 analisa a evolução da população indígena brasileira entre 2010 e 2022, comparando as mudanças demográficas ocorridas ao longo desse período sob três perspectivas: crescimento populacional, situação do domicílio e localização territorial.

A análise busca compreender não apenas a magnitude do crescimento da população indígena, mas também como sua distribuição entre áreas urbanas e rurais e dentro e fora de Terras Indígenas se modificou entre os dois censos.

## Objetivo e perguntas analíticas

A pergunta central deste estudo é:

> **Como evoluiu a população indígena brasileira entre 2010 e 2022?**

Para respondê-la, serão examinadas três dimensões:

1. **Crescimento populacional:** qual foi a magnitude do aumento da população indígena no período?
2. **Situação do domicílio:** como se modificou a distribuição da população indígena entre áreas urbanas e rurais?
3. **Localização territorial:** como evoluiu a população residente dentro e fora de Terras Indígenas?

Os resultados apresentados são de natureza descritiva. As variações observadas entre os censos de 2010 e 2022 não serão interpretadas automaticamente como evidência de migração, deslocamento populacional ou relação causal.

```python
from pathlib import Path

import pandas as pd

from charts.national_demographic_evolution import PlotBar
```

## 1. Carregamento da base de dados


```python
# Sobe as pastas até achar a raiz do projeto
ROOT = next(p for p in Path.cwd().parents if (p / "pyproject.toml").exists())

# Caminhdo para o arquivo
df_path = (
    ROOT
    / "data"
    / "processed"
    / "table"
    / "df_pais.csv"
)

# Caminho de saída
output_path = (
    ROOT
    / "outputs"
    / "figures"
    / "charts"
)

df_pais = pd.read_csv(df_path)  

grafico = PlotBar(df_pais, output_path=output_path)
```

## 2. O crescimento da população indígena entre 2010 e 2022 foi realmente significativo?

O primeiro passo consiste em comparar o contingente registrado nos dois últimos Censos Demográficos para verificar a magnitude da variação ocorrida ao longo do período.

- **Diferença Absoluta**

$$
\text{Diferença Absoluta} = V_f - V_i
$$


- **Crescimento Percentual**

$$
C\% = \frac{V_f - V_i}{V_i} \times 100
$$

```python
grafico.plot_crescimento_pais()
```

## Resposta

O gráfico evidencia que a população indígena registrada no Brasil apresentou um crescimento expressivo entre 2010 e 2022. Nesse intervalo, o contingente passou de aproximadamente 897 mil para quase 1,7 milhão de pessoas, representando um aumento de 89%.

Entretanto, esse resultado não deve ser interpretado exclusivamente como consequência do crescimento natural da população. Segundo o IBGE, a expansão observada também está associada ao aperfeiçoamento da metodologia censitária, à ampliação da cobertura da coleta em Terras Indígenas e ao aumento da autodeclaração, especialmente entre pessoas residentes fora desses territórios.

Sob essa perspectiva, o Censo de 2022 representa não apenas um aumento quantitativo da população registrada, mas também uma transformação na forma como a população indígena foi identificada e contabilizada, consolidando um novo patamar demográfico para esse grupo no Brasil.

## 3. Esse crescimento ocorreu da mesma forma nas áreas urbanas e rurais?

```python
grafico.plot_crescimento_area()
```

## Resposta
O gráfico mostra que não. Embora a população indígena tenha aumentado tanto nas áreas urbanas quanto nas rurais entre 2010 e 2022, o crescimento foi significativamente mais intenso nas cidades. Enquanto a população urbana aumentou 181,6%, a população rural cresceu 36,4% no mesmo período, alterando a composição da população indígena registrada no país.

Esse comportamento indica que o crescimento observado não pode ser atribuído apenas ao aumento natural da população. Segundo o IBGE, fatores como o aperfeiçoamento da metodologia censitária, a ampliação da cobertura da coleta e o aumento da autodeclaração indígena contribuíram para esse resultado, especialmente entre pessoas residentes em áreas urbanas. O fortalecimento da identidade étnica e a maior visibilidade dos povos indígenas também podem ter favorecido esse processo.

Ao mesmo tempo, o crescimento da população rural demonstra que as comunidades localizadas em áreas rurais e Terras Indígenas permaneceram em expansão. Assim, o fenômeno observado não representa apenas um deslocamento da população para as cidades, mas uma combinação de fatores demográficos, metodológicos e sociais.

Por fim, os resultados contribuem para desconstruir a percepção de que a população indígena está restrita às aldeias. O Censo de 2022 evidencia uma presença indígena cada vez mais significativa também nos centros urbanos, ampliando a necessidade de políticas públicas que contemplem essa realidade.

## 4. Onde a população indígena está concentrada: dentro ou fora das Terras Indígenas?

```python
grafico.plot_distribuicao_detalhada_ti()
```

## Resposta

O gráfico evidencia que, em 2022, a maior parte da população indígena registrada passou a viver fora das Terras Indígenas. Enquanto o contingente residente nesses territórios aumentou de aproximadamente 517 mil para 623 mil pessoas, a população localizada fora das Terras Indígenas cresceu de cerca de 380 mil para mais de 1,07 milhão de pessoas, tornando-se o grupo predominante.

Esse resultado representa uma mudança importante na distribuição da população indígena registrada entre os dois censos. Embora a população residente em Terras Indígenas tenha continuado a crescer, a expansão ocorreu de forma muito mais intensa fora desses territórios, especialmente em áreas urbanas, que concentraram o maior aumento absoluto observado no período.

É importante destacar que viver fora de uma Terra Indígena não significa deixar de ser indígena. Muitos povos indígenas residem em cidades ou em áreas rurais não demarcadas, mantendo vínculos culturais, comunitários e identitários com seus povos de origem. Segundo o IBGE, o crescimento observado fora das Terras Indígenas está associado, entre outros fatores, ao aperfeiçoamento da coleta censitária e ao aumento da autodeclaração indígena, além de processos sociais e demográficos ocorridos ao longo do período.

Os resultados reforçam que a realidade contemporânea da população indígena brasileira é mais diversa do que a percepção frequentemente associada apenas às Terras Indígenas. Essa transformação amplia a necessidade de políticas públicas que contemplem não apenas as comunidades localizadas em territórios demarcados, mas também a crescente população indígena residente fora dessas áreas, especialmente nos centros urbanos.


## Conclusão

A análise dos Censos Demográficos de 2010 e 2022 evidencia uma transformação importante na distribuição da população indígena registrada no Brasil. No período, o contingente praticamente dobrou, passando de aproximadamente 897 mil para quase 1,7 milhão de pessoas.

Os resultados mostram que esse crescimento ocorreu de forma desigual entre os diferentes contextos analisados. A expansão foi mais intensa nas áreas urbanas e concentrou-se principalmente fora das Terras Indígenas, indicando uma mudança relevante na distribuição espacial dessa população.

Esses resultados estão alinhados às interpretações apresentadas pelo IBGE, segundo as quais o aumento observado decorre da combinação entre o crescimento demográfico, o aperfeiçoamento da metodologia censitária e o fortalecimento da autodeclaração indígena.

Por fim, a análise reforça a importância de compreender a diversidade da realidade indígena contemporânea. A população indígena brasileira não está restrita às Terras Indígenas nem às áreas rurais, o que evidencia a necessidade de políticas públicas capazes de atender tanto as comunidades localizadas em territórios demarcados quanto a crescente população residente em áreas urbanas e fora desses territórios.
