# Preparação da Base Geográfica

Antes de realizar análises espaciais, é necessário integrar a base analítica construída na etapa de pré-processamento com a malha territorial oficial do IBGE. Essa integração permitirá representar espacialmente as informações sobre a população indígena brasileira e servirá de base para todas as visualizações e análises desenvolvidas nas etapas seguintes.

```python
from pathlib import Path

from geospatial.geographic import (
    PrepararBaseGeografica,
)
```

```python
ROOT = Path.cwd().parent

caminho_malha = (
    ROOT
    / "data"
    / "raw"
    / "BR_UF_2025"
    / "BR_UF_2025.shp"
)

caminho_base_estados = (
    ROOT
    / "data"
    / "processed"
    / "db_estados.csv"
)

diretorio_saida = ROOT / "data" / "geo"

caminho_malha.exists(), caminho_base_estados.exists(), diretorio_saida.exists()
```

```python
preparador = PrepararBaseGeografica(
    caminho_malha=caminho_malha,
    caminho_base_estados=caminho_base_estados,
)
```

```python
gdf_estados = preparador.executar(
    diretorio_saida=diretorio_saida
)
```
